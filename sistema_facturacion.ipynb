# Reglas de descuentos y envío

*   Si el valor de la compra antes de IVA es mayor o igual a $100.000, el sistema aplica automáticamente un 10 % de descuento.

*   Si el cliente compra 5 o más unidades del mismo producto, el sistema aplica automáticamente un 5 % de descuento sobre ese producto.

*   Si el total de la compra después de aplicar los descuentos es mayor o igual a $150.000, el sistema aplica automáticamente envío gratis.

Si el total es menor a $150.000, se cobra un costo de envío de $8.000.

```python
CATALOGO = [

    {"codigo": "P-001", "nombre": "Concentrado para perro 2 kg", "precio": 45000, "iva": 0.19},

    {"codigo": "P-002", "nombre": "Arena para gato 5 kg", "precio": 32000, "iva": 0.19},

    {"codigo": "P-003", "nombre": "Collar para perro", "precio": 25000, "iva": 0.19},

    {"codigo": "P-004", "nombre": "Shampoo para mascotas", "precio": 18000, "iva": 0.19},

    {"codigo": "P-005", "nombre": "Alimento húmedo para gato", "precio": 8000, "iva": 0.05},

    {"codigo": "P-006", "nombre": "Medicamento veterinario", "precio": 20000, "iva": 0.0},

]

numero_factura = 1000
```

```python
def mostrar_catalogo():

    print("\nCATÁLOGO DISPONIBLE")

    print("-" * 65)

    for p in CATALOGO:

        print(f"{p['codigo']:<8} {p['nombre']:<32} ${p['precio']:<10} IVA: {p['iva']*100:.0f}%")

    print("-" * 65)


def buscar_producto(codigo):

    for p in CATALOGO:

        if p["codigo"].upper() == codigo.upper():

            return p

    return None


def calcular_descuento(subtotal, cantidad):

    descuento = 0.0

    # Regla 1: compra antes de IVA >= $100.000
    if subtotal >= 100000:

        descuento += 0.10

    # Regla 2: 5 o más unidades del mismo producto
    if cantidad >= 5:

        descuento += 0.05

    return descuento


mostrar_catalogo()
```

```python
def tomar_pedido():

    items_factura = []

    print("Escribe el código del producto (ej: P-001). Escribe 'fin' para terminar.")

    while True:

        codigo = input("Código de producto: ").strip()

        if codigo.lower() == "fin":

            break

        producto = buscar_producto(codigo)

        if producto is None:

            print("⚠️ Producto no encontrado. Intenta de nuevo.")

            continue

        cantidad_str = input(f"Cantidad de '{producto['nombre']}': ").strip()

        if not cantidad_str.isdigit() or int(cantidad_str) <= 0:

            print("⚠️ Cantidad inválida. Intenta de nuevo.")

            continue

        cantidad = int(cantidad_str)

        subtotal = producto["precio"] * cantidad

        items_factura.append({

            "codigo": producto["codigo"],
            "nombre": producto["nombre"],
            "precio": producto["precio"],
            "cantidad": cantidad,
            "iva": producto["iva"],
            "subtotal": subtotal

        })

        print(f"✅ Agregado: {cantidad} x {producto['nombre']} = ${subtotal}")

    return items_factura
```

```python
def generar_factura(items_factura, nombre_cliente):
    global numero_factura

    if len(items_factura) == 0:
        print("No hay productos en la factura.")
        return

    # 1. Calcular subtotal de toda la compra
    subtotal = 0

    for item in items_factura:
        subtotal += item["subtotal"]

    # 2. Regla 1: descuento del 10% si la compra
    # antes de IVA es mayor o igual a $100.000
    if subtotal >= 100000:
        descuento_compra = subtotal * 0.10
    else:
        descuento_compra = 0

    # 3. Regla 2: descuento del 5% si compra
    # 5 o más unidades del mismo producto
    descuento_cantidad = 0

    for item in items_factura:
        if item["cantidad"] >= 5:
            descuento_cantidad += item["subtotal"] * 0.05

    # 4. Total de descuentos
    valor_descuento = descuento_compra + descuento_cantidad

    # 5. Base después de descuentos
    base_despues_descuentos = subtotal - valor_descuento

    # 6. Calcular IVA de cada producto
    # teniendo en cuenta su porcentaje de IVA
    valor_iva = 0

    for item in items_factura:

        # Participación del producto en el descuento general
        if subtotal > 0:
            descuento_general_item = (
                item["subtotal"] / subtotal
            ) * descuento_compra
        else:
            descuento_general_item = 0

        # Descuento por cantidad para ese producto
        if item["cantidad"] >= 5:
            descuento_cantidad_item = item["subtotal"] * 0.05
        else:
            descuento_cantidad_item = 0

        # Base del producto después de descuentos
        base_item = (
            item["subtotal"]
            - descuento_general_item
            - descuento_cantidad_item
        )

        # IVA según el producto
        iva_item = base_item * item["iva"]

        valor_iva += iva_item

    # 7. Total después de descuentos e IVA
    total_antes_envio = base_despues_descuentos + valor_iva

    # 8. Regla 3: envío gratis desde $150.000
    if total_antes_envio >= 150000:
        costo_envio = 0
    else:
        costo_envio = 8000

    # 9. Total final
    total_final = total_antes_envio + costo_envio

    # 10. Aumentar número de factura
    numero_factura += 1

    # 11. Mostrar factura
    print("\n" + "=" * 60)
    print("              PATITAS PET SHOP")
    print("              FACTURA DE VENTA")
    print("=" * 60)

    print(f"Factura N.º: {numero_factura}")
    print(f"Cliente: {nombre_cliente}")

    print("-" * 60)
    print("DETALLE DE LA COMPRA")
    print("-" * 60)

    for item in items_factura:
        print(
            f"{item['codigo']} - {item['nombre']}"
        )
        print(
            f"Cantidad: {item['cantidad']} x ${item['precio']}"
        )
        print(
            f"Subtotal: ${item['subtotal']}"
        )
        print(
            f"IVA: {item['iva'] * 100:.0f}%"
        )
        print("-" * 60)

    print(f"Subtotal: ${subtotal:.0f}")
    print(f"Descuento: ${valor_descuento:.0f}")
    print(f"IVA: ${valor_iva:.0f}")
    print(f"Envío: ${costo_envio:.0f}")
    print("=" * 60)
    print(f"TOTAL A PAGAR: ${total_final:.0f}")
    print("=" * 60)

    # 12. Guardar la factura también como archivo .txt
    guardar_factura_txt(
        numero_factura,
        nombre_cliente,
        items_factura,
        subtotal,
        valor_descuento,
        valor_iva,
        costo_envio,
        total_final
    )


    return total_final

```

```python
#Hacer que la factura se guarde en .txt

def guardar_factura_txt(numero, nombre_cliente, items_factura,
                        subtotal, valor_descuento, valor_iva,
                        costo_envio, total_final):

    nombre_archivo = f"factura_{numero}.txt"

    with open(nombre_archivo, "w", encoding="utf-8") as archivo:

        archivo.write("=" * 60 + "\n")
        archivo.write("              PATITAS PET SHOP\n")
        archivo.write("              FACTURA DE VENTA\n")
        archivo.write("=" * 60 + "\n")

        archivo.write(f"Factura N.º: {numero}\n")
        archivo.write(f"Cliente: {nombre_cliente}\n")

        archivo.write("-" * 60 + "\n")
        archivo.write("DETALLE DE LA COMPRA\n")
        archivo.write("-" * 60 + "\n")

        for item in items_factura:
            archivo.write(
                f"{item['codigo']} - {item['nombre']}\n"
            )
            archivo.write(
                f"Cantidad: {item['cantidad']} x ${item['precio']}\n"
            )
            archivo.write(
                f"Subtotal: ${item['subtotal']:.0f}\n"
            )
            archivo.write(
                f"IVA: {item['iva'] * 100:.0f}%\n"
            )
            archivo.write("-" * 60 + "\n")

        archivo.write(f"Subtotal: ${subtotal:.0f}\n")
        archivo.write(f"Descuento: ${valor_descuento:.0f}\n")
        archivo.write(f"IVA: ${valor_iva:.0f}\n")
        archivo.write(f"Envío: ${costo_envio:.0f}\n")
        archivo.write("=" * 60 + "\n")
        archivo.write(f"TOTAL A PAGAR: ${total_final:.0f}\n")
        archivo.write("=" * 60 + "\n")

    print(f"Archivo generado: {nombre_archivo}")
```

```python
def sistema_facturacion():
    print("BIENVENIDO AL SISTEMA DE FACTURACIÓN")
    print("HUELLITAS PET SHOP")

    mostrar_catalogo()

    nombre_cliente = input("Nombre del cliente: ").strip()

    items = tomar_pedido()

    generar_factura(items, nombre_cliente)


# Descomenta para ejecutar el sistema completo:
sistema_facturacion()
```
